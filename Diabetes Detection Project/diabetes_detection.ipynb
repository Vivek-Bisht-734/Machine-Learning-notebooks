```python
import tensorflow as tf
import keras
import numpy as np
import scipy
import h5py

from numpy import loadtxt
from keras.models import Sequential, model_from_json
from keras.layers import Input, Dense

print(tensorflow.__version__)  # 1.14.0
print(keras.__version__)  # 2.2.4
print(np.__version__)  # 1.18.5
print(scipy.__version__)  # 1.1.0
print(h5py.__version__)  # 2.10.0
```

```python
# In the Dataset
"""
1. Number of times pregnant
2. Plasma glucose concentration a 2 hours in an oral glucose tolerance test
3. Diastolic blood pressure (mm Hg)
4. Triceps skin fold thickness (mm)
5. 2-Hour serum insulin (mu U/ml)
6. Body mass index (weight in kg / (height in m)^2)
7. Diabestes pedigree function
8. Age (years)
9. Class variable (0 or 1)
"""
```

```python
# Loading Dataset
dataset = loadtxt("pima-indians-diabetes.csv", delimiter = ',')
x = dataset[:, :8]
y = dataset[:, 8]
print(x)
print(y)
```

```python
# Training
model = Sequential()
model.add(Input(shape = (8,)))
model.add(Dense(12, activation = 'relu'))
model.add(Dense(8, activation = 'relu'))
model.add(Dense(1, activation = 'sigmoid'))

model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
model.fit(x, y, epochs = 30, batch_size = 10)

_, accuracy = model.evaluate(x, y)
print('Accuracy: %.2f' % (accuracy * 100))

model_json = model.to_json()
with open('model.json', 'w') as json_file:
    json_file.write(model_json)
model.save_weights('model.weights.h5')
print('Saved model to disk')
```

```python
# Testing
dataset = loadtxt('pima-indians-diabetes.csv', delimiter = ',')
x = dataset[:, :8]
y = dataset[:, 8]

json_file = open('model.json', 'r')
loaded_model_json = json_file.read()
json_file.close()

model = model_from_json(loaded_model_json)
model.load_weights('model.weights.h5')
print('Loaded model from disk')

probabilities = model.predict(x)
predictions = (probabilities >= 0.5).astype(int)

for i in range(5, 10):
    print('%s => %d (Original Class: %d)' % (x[i].tolist(), predictions[i][0], y[i]))
```
